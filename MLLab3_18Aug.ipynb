```python
#Split independent variables
X = dataset.iloc[:,:-1]
X

#Split dependent variables
y = dataset.iloc[:, 3].values
y
```

```python
#Taking care of missing data
from sklearn.impute import SimpleImputer
import numpy as np
imputer = SimpleImputer(missing_values=np.nan, strategy='mean') #Object of Simple Imputer Class

imputer = imputer.fit(X.iloc[:,1:3]) # Used to calculate the mean value for columns  1 and 3
X.iloc[:, 1:3] = imputer.transform(X.iloc[:, 1:3]) # Fills missing value
X
```
