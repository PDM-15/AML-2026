```python
people = ['Python programming Language', 'Computer lab test', 'Vinod scored A']

def split_title_and_name(person):
  title = person.split()[0]
  lname = person.split()[-1]

  return '{} {}'.format(title, lname)

print(list(map(split_title_and_name,people)))
```
