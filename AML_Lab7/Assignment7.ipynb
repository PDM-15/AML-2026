```python
from google.colab import drive

drive.mount('/content/drive')
import pandas as pd

# Update the path below to match your actual file location
file_path = '/content/drive/My Drive/Dataset/data-1.csv'
# View the first few rows
df = pd.read_csv(file_path)
df.head()
```

```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold

X = df

selector = VarianceThreshold(threshold = 0.0)
X_selected = selector.fit_transform(X)

selected_columns = X.columns[selector.get_support()]

print("Selected Features:", list(selected_columns))
```

```python
from google.colab import drive

drive.mount('/content/drive')
import pandas as pd

# Update the path below to match your actual file location
file_path = '/content/drive/My Drive/Dataset/data-2.csv'
# View the first few rows
df2 = pd.read_csv(file_path)
df2.head()
```
