Random Forest

```python
import numpy as np
import pandas as pd
from google.colab import drive
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

#Loading
drive.mount('/content/drive')

data = pd.read_csv("drive/My Drive/Dataset/Social_Network_Ads.csv")
dataset = data

data.info()

X = dataset.iloc[:,[2,3]].values

y = dataset.iloc[:,4].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 45)

sc_X = StandardScaler()
X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, classification_report
from sklearn import metrics
from sklearn import tree

classifier = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 45) #We can give gini as criterion as well
classifier.fit(X_train, y_train)

y_pred = classifier.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

print("Classification Report:")
print(classification_report(y_test, y_pred))

prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:", prediction * 100, "%")

new_individual = np.array([[29, 80000]])
new_individual_scaled = sc_X.transform(new_individual)
prediction = classifier.predict(new_individual_scaled)

if prediction[0] == 1:
  print("User will buy")
else:
  print("User will not buy")
```

K Nearest Neighbours

```python
import numpy as np
import pandas as pd
from google.colab import drive
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

#Loading
drive.mount('/content/drive')

data = pd.read_csv("drive/My Drive/Dataset/Social_Network_Ads.csv")
dataset = data

data.info()

X = dataset.iloc[:,[2,3]].values

y = dataset.iloc[:,4].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 45)

sc_X = StandardScaler()
X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, classification_report
from sklearn import metrics
from sklearn import tree

classifier = KNeighborsClassifier(n_neighbors = 5, metric = 'minkowski', p = 2)
classifier.fit(X_train, y_train)

y_pred = classifier.predict(X_test)
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

print("Classification Report:")
print(classification_report(y_test, y_pred))

prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:", prediction * 100, "%")
```

AdaBoost

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import confusion_matrix, classification_report
from sklearn import metrics
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

drive.mount('/content/drive')

data = pd.read_csv("drive/My Drive/Dataset/mushrooms.csv")
for label in data.columns:
  data[label] = LabelEncoder().fit(data[label]).transform(data[label])

dataset = data

X = dataset.iloc[:,1:].values
y = dataset.iloc[:,0].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 45)

model = DecisionTreeClassifier(criterion = 'entropy', max_depth = 1)

adaboost = AdaBoostClassifier(estimator = model, n_estimators = 400, learning_rate = 1.0, random_state = 45)
boostmodel = adaboost.fit(X_train, y_train)

y_pred = boostmodel.predict(X_test)
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

print("Classification Report:")
print(classification_report(y_test, y_pred))

prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:", prediction * 100, "%")
```

PCA

```python
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report
from sklearn import metrics

drive.mount('/content/drive')

data = pd.read_csv("drive/My Drive/Dataset/Wine.csv")
dataset = data

X = dataset.iloc[:,0:13].values
y = dataset.iloc[:,13].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 45)

sc_X = StandardScaler()

X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)

pca = PCA(n_components = 2)
X_train = pca.fit_transform(X_train)
X_test = pca.transform(X_test)

explained_variance = pca.explained_variance_ratio_
print(explained_variance)

classifier = LogisticRegression(random_state = 45)
classifier.fit(X_train, y_train)

y_pred = classifier.predict(X_test)
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

print("Classification Report:")
print(classification_report(y_test, y_pred))

prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:", prediction * 100, "%")
```
