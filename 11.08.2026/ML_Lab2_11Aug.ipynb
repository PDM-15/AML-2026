```python
x = np.array([1,2,3])
y = np.array([1,4,6])

print(x+y) #elementwise addition of arrays
print(x-y) #elementwise subtraction of arrays
print(x*y) #elementwise multiplication of elements
print(x@y) #elementwise addition of the product array
print(x/y) #elementwise division of array
print(x**2) #elementwise square of the array
print(x.dot(y)) #dot product

```

```python
a = np.arange(100)
a.resize(10,10) #resize into desired no. of rows and columns

a[3,4] #rows and columns
a[2,1:4]
a[3,3:6]
a[:3,:-1]
a[-1, ::2] #row columns and steps
a[a>30] #elements greater than 30
a[a>50] = 50 # assigning particular range of values into a value
```

```python
a2 = a[:4, :4] #instance of a
a2[:] = 0 #overwrites values in a

a_copy = a.copy() # to overcome the overwrite problem use copy
a_copy[:] = 10 #even if modified, doesnt overwrite the original
```

```python
test = np.random.randint(0,10,(4,3))

for row in test: #used to loop through the rows
  print(row)

for i in range(len(test)):
  print(test[i])

for i, row in enumerate(test): #iterate by row and index
  print('row', i, 'is', row)

```
