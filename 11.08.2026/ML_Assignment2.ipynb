23. Create a NumPy array containing 100 random exam marks and determine how many students fall into the ranges 0–39, 40–59, 60–79, and 80–100

24. Compare the time required to multiply 1,000,000 numbers using a Python list and a NumPy array

```python
import time
import numpy as np

# Number of elements (1 million)
size = 1_000_000

# --- 1. Python List Approach ---
# Creating two lists of 1 million items
list_a = list(range(size))
list_b = list(range(size))

start_time = time.time()
# Element-wise multiplication using a list comprehension
list_result = [a * b for a, b in zip(list_a, list_b)]
list_duration = time.time() - start_time

print(f"Python list time: {list_duration:.6f} seconds")


# --- 2. NumPy Array Approach ---
# Creating two NumPy arrays of 1 million items
array_a = np.arange(size)
array_b = np.arange(size)

start_time = time.time()
# Element-wise multiplication using a vectorized operation
array_result = array_a * array_b
array_duration = time.time() - start_time

print(f"NumPy array time: {array_duration:.6f} seconds")


# --- 3. Speed Factor ---
speedup = list_duration / array_duration
print(f"NumPy is approximately {speedup:.1f}x faster!")
```
