10. Assign grades according to:

	| Marks | Grade |
	| ----: | ----- |
	|  ≥ 90 | A     |
	|  ≥ 80 | B     |
	|  ≥ 70 | C     |
	|  < 70 | D     |


```python
for i in data.index:
  score = data.loc[i, "Marks"]
  if score >= 90:
    data.loc[i, "Grade"] = "A"
  elif score >= 80 and score <= 90:
    data.loc[i, "Grade"] = "B"
  elif score >= 70 and score <= 80:
    data.loc[i, "Grade"] = "C"
  else:
    data.loc[i, "Grade"] = "D"

print(data[["Name", "Marks", "Grade"]])
```

19. Use the following dataset:

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)


Perform the following analysis:

Display the first five records.
Find the dimensions of the dataset.
Find the average final mark.
Find the highest-scoring student.
Find the lowest-scoring student.
Display students scoring above 80.
Find average marks by department.
Find students having attendance above 80%.
Sort students by FinalMarks.
Add a Result column: Pass if FinalMarks ≥ 70, otherwise Fail.
Add a performance category:
	≥90 → Excellent
	≥80 → Very Good
	≥70 → Good
	<70 → Needs Improvement
Find the department having the highest average FinalMarks.


```python
import pandas as pd
data = { "Name": [ "Asha", "Ravi", "Meena", "Kiran", "Arun", "Divya", "John", "Priya" ],
         "Department": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE", "ME", "ECE"],
         "Attendance": [92, 75, 96, 68, 82, 88, 72, 95 ],
         "Internal": [85, 72, 94, 65, 80, 86, 70, 91],
         "FinalMarks": [88, 74, 96, 62, 83, 89, 68, 93]}

df = pd.DataFrame(data)

print(df[:5])
print("-----------------------------------------")
print(df.shape)
print("-----------------------------------------")
print(df["FinalMarks"].mean())
print("-----------------------------------------")
print(df["FinalMarks"].max())
print("-----------------------------------------")
print(df["FinalMarks"].min())
print("-----------------------------------------")
print(df[df["FinalMarks"] > 80])
print("-----------------------------------------")
print(df.groupby("Department")["FinalMarks"].mean())
print("-----------------------------------------")
print(df[df["Attendance"] > 80])
print("-----------------------------------------")
print(df.sort_values(by = "FinalMarks"))
print("-----------------------------------------")
df["Result"] = df["FinalMarks"].apply(lambda x : x >= 70 and "Pass" or "Fail")
print(df)
print("-----------------------------------------")
df["Performance"] = df["FinalMarks"].apply(lambda x : x >= 90 and "Excellent" or x >= 80 and "Very Good" or x >= 70 and "Good" or "Needs Improvement")
print(df)
```

20. Create:

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5

    "Salary": [

    ]

Determine:

Average salary.
Maximum and minimum salary.
Employee receiving the highest salary.
Employees earning above ₹60,000.
Average salary for each department.
Average experience by department.
Number of employees in each department.
Sort employees by salary.
Find IT employees with more than 3 years of experience.
Add a SalaryLevel column:
	≥ ₹75,000 → High
	₹50,000–₹74,999 → Medium
	< ₹50,000 → Low

```python
employees = pd.DataFrame({ "Employee": [ "E01", "E02", "E03", "E04", "E05", "E06", "E07", "E08" ],
                           "Department": [ "IT", "HR", "IT", "Finance", "HR", "Finance", "IT", "Finance"],
                           "Experience": [5, 3, 7, 4, 6, 8, 2, 5],
                           "Salary": [60000, 45000, 75000, 65000, 55000, 85000, 48000, 70000]})
print(employees["Salary"].mean())
print("-----------------------------------------")
print(employees["Salary"].max())
print(employees["Salary"].min())
print("-----------------------------------------")
print(employees[employees["Salary"] == employees["Salary"].max()])
print("-----------------------------------------")
print(employees[employees["Salary"] > 60000])
print("-----------------------------------------")
print(employees.groupby("Department")["Salary"].mean())
print("-----------------------------------------")
print(employees.groupby("Department")["Experience"].mean())
print("-----------------------------------------")
print(employees.groupby("Department")["Employee"].sum())
print("-----------------------------------------")
print(employees.sort_values(by = "Salary"))
print("-----------------------------------------")
print(employees[(employees["Department"] == 'IT') & (employees["Experience"] > 3)])
print("-----------------------------------------")
employees["SalaryLevel"] = employees["Salary"].apply(lambda x : x >= 75000 and "High" or x >= 50000 and "Medium" or "Low")
print(employees)
#
```
